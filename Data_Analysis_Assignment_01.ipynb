## 1. Enter a three-digit number and find the sum of its digits

The number is read as an integer. I separate the hundreds, tens and units digits and then add them.


```python
number = int(input("Enter a three-digit number: "))

if 100 <= number <= 999:
    hundreds = number // 100
    tens = (number // 10) % 10
    units = number % 10

    digit_sum = hundreds + tens + units
    print("Sum of the digits:", digit_sum)
else:
    print("Please enter a three-digit number.")
```

## 2. Compute the given formulas

For the first formula, I use the expression exactly as given in the task:

`y = -(b + sqrt(b² - 4ac)) / (2a)`

For the second formula:

`y = sin²(x) + log₂(1 + |x|)`

The trigonometric function in Python uses radians.


```python
import math

a = float(input("Enter a: "))
b = float(input("Enter b: "))
c = float(input("Enter c: "))
x = float(input("Enter x (in radians): "))

discriminant = b**2 - 4*a*c

if a != 0 and discriminant >= 0:
    y1 = -(b + math.sqrt(discriminant)) / (2*a)
    print("First formula:", y1)
else:
    print("The first formula cannot be evaluated with these values.")

if 1 + abs(x) > 0:
    y2 = math.sin(x)**2 + math.log2(1 + abs(x))
    print("Second formula:", y2)
```

## 3. Guess the number game

The computer chooses a random number from 1 to 100. The player keeps guessing until the correct number is entered.


```python
import random

secret_number = random.randint(1, 100)
attempts = 0

print("I have chosen a number between 1 and 100.")

while True:
    guess = int(input("Your guess: "))
    attempts += 1

    if guess < secret_number:
        print("Too low.")
    elif guess > secret_number:
        print("Too high.")
    else:
        print(f"Correct! You found it in {attempts} attempt(s).")
        break
```

## 6. Function returning the sum and product of three numbers

The function returns two values: the sum and the product.


```python
def sum_and_product(a, b, c):
    total = a + b + c
    product = a * b * c
    return total, product


x, y, z = 2, 3, 4

total, product = sum_and_product(x, y, z)

print("Sum:", total)
print("Product:", product)
```

## 7. Reverse a string and remove the first character using slices

Python slicing makes both operations simple. `[::-1]` reverses the string, while `[1:]` removes the first character.


```python
text = input("Enter a string: ")

reversed_text = text[::-1]
without_first = text[1:]

print("Reversed string:", reversed_text)
print("Without first character:", without_first)
```

## 8. Replace tabs/newlines with spaces and remove punctuation

I use `translate()` to remove punctuation and replace tab/newline characters with spaces. Finally, I remove extra spaces.


```python
import string

text = input("Enter text: ")

text = text.replace("\t", " ").replace("\n", " ")
text = text.translate(str.maketrans("", "", string.punctuation))
text = " ".join(text.split())

print("Cleaned text:", text)
```

## 9. Remove all maximum elements from a list

First I find the maximum value. Then I create a new list containing every element except that maximum value.


```python
numbers = [4, 8, 2, 8, 5, 8, 1]

maximum = max(numbers)
new_list = [number for number in numbers if number != maximum]

print("Original list:", numbers)
print("Maximum value:", maximum)
print("New list:", new_list)
```

## 10. Sort by the last digit

The `key` argument tells Python to use the last digit of each number for sorting.


```python
numbers = [23, 41, 15, 32, 58, 76, 14]

ascending = sorted(numbers, key=lambda number: number % 10)
descending = sorted(numbers, key=lambda number: number % 10, reverse=True)

print("Ascending by last digit:", ascending)
print("Descending by last digit:", descending)
```

## 12. Count characters in a string using a dictionary

Each character is used as a dictionary key, and its number of occurrences is stored as the value.


```python
text = input("Enter a string: ")

character_count = {}

for character in text:
    character_count[character] = character_count.get(character, 0) + 1

print("Character counts:")
print(character_count)
```

## 13. Find a word starting with a capital letter and length less than 5

The regular expression looks for a word that starts with an uppercase English letter and has a total length of 1 to 4 letters.


```python
import re

text = input("Enter text: ")

pattern = r"\b[A-Z][A-Za-z]{0,3}\b"
match = re.search(pattern, text)

if match:
    print("First matching word:", match.group())
else:
    print("No matching word was found.")
```

## 14. Extract last name, first name and patronymic using regex

The expected format is similar to:

`Full Name: Ivanov Ivan Ivanovich`

The regular expression extracts the three name parts after `Full Name:`.


```python
import re

text = input("Enter a full name in the form 'Full Name: Last First Patronymic': ")

pattern = r"^Full Name:\s*([A-Za-zА-Яа-яЁё-]+)\s+([A-Za-zА-Яа-яЁё-]+)\s+([A-Za-zА-Яа-яЁё-]+)\s*$"
match = re.match(pattern, text)

if match:
    last_name, first_name, patronymic = match.groups()

    print("Last name:", last_name)
    print("First name:", first_name)
    print("Patronymic:", patronymic)
else:
    print("The name does not match the expected format.")
```
