## 1. Enter a three-digit number and find the sum of its digits.

The number is read as an integer. I separate the hundreds, tens and units digits and then add them.


```python
number = int(input("Enter a three-digit number: "))

if 100 <= number <= 999:
    hundreds = number // 100
    tens = (number // 10) % 10
    units = number % 10

    digit_sum = hundreds + tens + units
    print("Sum of the digits:", digit_sum)
else:
    print("Please enter a three-digit number.")
```

## Conclusion

The process involves dividing and mod-ing each digit and then adds them together.

## 2. Compute Formulas:

```python
import math

a = float(input("Enter a: "))
b = float(input("Enter b: "))
c = float(input("Enter c: "))

discriminant = b**2 - 4*a*c

if discriminant >= 0:
    y1 = (-b + math.sqrt(discriminant)) / (2*a)
    print("Result of first formula =", y1)
else:
    print("No real solution")

x = float(input("Enter x: "))

y2 = math.sin(x)**2 + math.log2(abs(1 + x))

print("Result of second formula =", y2)
```

## Conclusion

The first formula uses the quadratic formula using the discriminant method, while the second formula combines a trigonometric function with a logarithmic function.

## 3. "Guess the number" game.

A random number between 1 and 100 was chosen by the computer. The player keeps guessing until the correct number is entered.


```python
import random

number = random.randint(1, 100)

while True:
    try:
        guess = int(input("Guess the number (1-100): "))
    except ValueError:
        print("Invalid input. Please enter a whole number.")
        continue

    if guess == number:
        print("Congratulations! You guessed the correct number.")
        break
    elif guess < number:
        print("The number is higher.")
    else:
        print("The number is lower.")
```

## Conclusion

The game ends only when the number is correctly identified.

## 5. Count four-digit numbers containing digit 0.

A four-digit number is from 1000 to 9999. I check every number and count the ones that contain `0`.


```python
count = 0

for number in range(1000, 10000):
    if "0" in str(number):
        count += 1

print("Number of four-digit numbers containing 0:", count)
```

## 8. Replace tabs/newlines with spaces and remove punctuation.

I use `translate()` to remove punctuation and replace tab/newline characters with spaces. Finally, I remove extra spaces.


```python
import string

text = input("Enter text: ")

text = text.replace("\t", " ").replace("\n", " ")
text = text.translate(str.maketrans("", "", string.punctuation))
text = " ".join(text.split())

print("Cleaned text:", text)
```

## Conclusion

The text becomes cleaner and easier to process.

## 9. Remove all max elements from a list (new list).

First I find the maximum value. Then I create a new list containing every element except that maximum value.


```python
numbers = [4, 8, 2, 8, 5, 8, 1]

maximum = max(numbers)
new_list = [number for number in numbers if number != maximum]

print("Original list:", numbers)
print("Maximum value:", maximum)
print("New list:", new_list)
```

## Conclusion

I find the maximum value in a list and remove all occurrences of it to create a new list.

## 10. Sort by (1) last digit ascending, (2) descending.


```python
numbers = [23, 41, 15, 32, 58, 76, 14]

ascending = sorted(numbers, key=lambda number: number % 10)
descending = sorted(numbers, key=lambda number: number % 10, reverse=True)

print("Ascending by last digit:", ascending)
print("Descending by last digit:", descending)
```

## 12. Count characters in a string using a dictionary.

Each character is used as a dictionary key, and its number of occurrences is stored as the value.


```python
text = input("Enter a string: ")

character_count = {}

for character in text:
    character_count[character] = character_count.get(character, 0) + 1

print("Character counts:")
print(character_count)
```

## Conclusion

I count the number of times each character appears in a string using a Python dictionary.

## 13. Find a word starting with capital letter and length < 5 (regex).

The regular expression looks for a word that starts with an uppercase English letter and has a total length of 1 to 4 letters.


```python
import re

text = input("Enter text: ")

pattern = r"\b[A-Z][A-Za-z]{0,3}\b"
match = re.search(pattern, text)

if match:
    print("First matching word:", match.group())
else:
    print("No matching word was found.")
```

## Conclusion

I used regular expressions in Python to find a word that matches a specific pattern.

## 14. Extract last name, first name and patronymic from "Full Name:..."(regex).

The expected format will be:

"Full Name: Prozorov Oleg Alexandrovich"

The regular expression extracts the three name parts after 'Full Name:'.


```python
import re

text = "Full Name: Prozorov Oleg Alexandrovich"

match = re.search(
    r'Full Name:\s+([A-Za-z]+)\s+([A-Za-z]+)\s+([A-Za-z]+)',
    text
)

if match:
    last_name = match.group(1)
    first_name = match.group(2)
    patronymic = match.group(3)

    print("Last Name:", last_name)
    print("First Name:", first_name)
    print("Patronymic:", patronymic)
```

## Conclusion

I used regular expressions to validate and separate a full name into its different parts.
